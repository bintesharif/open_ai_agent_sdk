```python
from agents import Agent, AsyncOpenAI, Runner, OpenAIChatCompletionsModel
from agents.run import RunConfig
from google.colab import userdata
from agents.extensions.visualization import draw_graph
import nest_asyncio
import asyncio

nest_asyncio.apply()

MODEL = 'gemini/gemini-2.0-flash'
API_KEY = userdata.get('GEMINI_API_KEY')


external_model = AsyncOpenAI(
    api_key=API_KEY,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",

)

LLM_MODEL = OpenAIChatCompletionsModel(
    model="gemini-2.0-flash",
    openai_client=external_model
)

config = RunConfig(
    model=LLM_MODEL,
    tracing_disabled=True
)
```

```python
web_developer_agent: Agent = Agent(
    name="Web Developer",
    instructions="You are Web_Developer, an expert assistant providing accurate web development answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.",
    model=LLM_MODEL,
    handoff_description="Web Developer expert"

)

APP_developer_agent: Agent = Agent(
    name="APP Developer",
    instructions="""You are APP_Developer, an expert assistant providing accurate APP development answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.""",
    model=LLM_MODEL,
    handoff_description="APP Developer expert"
)

# Define backend_developer_agent before using it
backend_developer_agent: Agent = Agent(
    name="Backend Developer",
    instructions="""You are Backend_Developer, an expert assistant providing accurate backend development answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.""",
    model=LLM_MODEL,
    handoff_description="Backend Developer expert"
)

# Define devops_developer_agent before using it tool
devops_developer_agent: Agent = Agent(
    name="DevOps Developer",
    instructions="""You are DevOps_Developer, an expert assistant providing accurate DevOps development answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.""",
    model=LLM_MODEL,
    handoff_description="DevOps Developer expert"
)


backend_developer_tool = backend_developer_agent.as_tool( # Use backend_developer_agent
    tool_name = "backend_developer",
    tool_description="A tool that provides expert backend development assistance, including API design, database management, server-side logic, and best practices for scalable systems."
)

devops_developer_tool = devops_developer_agent.as_tool( # Use devops_developer_agent
    tool_name = "devops_developer",
    tool_description="A tool that offers DevOps expertise, guiding on CI/CD pipelines, cloud infrastructure, containerization, and automation for efficient deployments."
)
Agentic_Ai_developer: Agent = Agent(
    name="Agentic_Ai_developer",
    instructions="""You are agentic_Ai_developer, an expert assistant providing accurate agentic_Ai_developer answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.""",
    model=LLM_MODEL,
    handoff_description="Agentic_Ai_developer expert",
    tools=[backend_developer_tool, devops_developer_tool]
)

panacloud_agent: Agent = Agent(
    name="panacloud_agent",
    instructions="""You are panacloud_agent, an expert assistant providing accurate panacloud_agent answers, generating clean code snippets, and suggesting best practices in a clear, professional tone.""",
    model=LLM_MODEL,
    handoffs=[web_developer_agent, APP_developer_agent, Agentic_Ai_developer]
)

output = Runner.run_sync(panacloud_agent, "what is web developer" )


print(output.last_agent)
print(output.final_output)
```
